```python
import sentencepiece as spm
from pathlib import Path
```

```python
corpus = """
machine learning is a field of artificial intelligence
machine learning algorithms learn patterns from data
deep learning is a subset of machine learning
learning algorithms can analyze data
machine learning models are trained on data
researchers use machine learning for image classification
"""

corpus_file = "corpus.txt"

Path(corpus_file).write_text(corpus.strip(), encoding="utf-8")
```

```python
spm.SentencePieceTrainer.train(
    input=corpus_file,
    model_prefix="my_bpe",
    vocab_size=100,
    model_type="bpe",
    character_coverage=1.0,
    pad_id=0,
    unk_id=1,
    bos_id=2,
    eos_id=3
)

print("BPE tokenizer trained successfully!")
```

```python
sp = spm.SentencePieceProcessor(
    model_file="my_bpe.model"
)
```

```python
vocab_size = sp.get_piece_size()
vocab_size
```

```python
for i in range(vocab_size):
    piece = sp.id_to_piece(i)
    print(f"{i} -> {piece}")
```

```python
text = "machine learning models learn new patterns"

pieces = sp.encode(
    text,
    out_type=str
)
```

```python
ids = sp.encode(
    text,
    out_type=int
)
```

```python
print("Tokens:",pieces)
print("IDs:",ids)
```

```python
for piece, token_id in zip(pieces, ids):
    print(f"{piece:15} -> {token_id}")
```

```python
decoded_text = sp.decode(ids)
```
